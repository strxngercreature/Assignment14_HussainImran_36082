Q5: Draw a bar chart showing average order value by city. 


```python
import matplotlib.pyplot as plt
```

```python
avg_city = ped.groupby("city")["order_value"].mean()
```

```python
categories = avg_city.index
values = avg_city.values
plt.bar(categories, values, color='lavender', width=0.5)
plt.xlabel('City')
plt.ylabel('Average Order Value')
plt.title('Average Order Value by City')
plt.show()
```

Dataset 2: Student Lifestyle & Study Habits 

```python
slh = pd.read_csv(r"C:\Users\Admin\Downloads\student_lifestyle_dataset.csv")
```

```python
slh.head()
```

```python
slh.columns
```

Q1: What is the average number of study hours?

```python
slh["study_hours"].mean()
```

Q2. Who sleeps more on average – males or females? 

```python
slh.groupby("gender")["sleep_hours"].mean().idxmax()
```

Q3: Is there a relationship between screen time and sleep hours? 

```python
slh["screen_time"].corr(slh["sleep_hours"])
```

As shown above, a correlation coefficient of +0.52 indicates a positive relationship between screen time and sleep hours. It suggests that students with higher screen time tend to have higher sleep hours as well.

Q4: Draw a histogram of daily transport cost.

```python
plt.hist(slh["daily_transport_cost"], bins=10, color='yellow', edgecolor='black', alpha=0.7)
plt.xlabel('Daily Transport Cost')
plt.ylabel('Frequency')
plt.title('Distribution of Daily Transport Cost')
plt.show()
```

Q5: Compare average internet usage by gender.

```python
avg_internet = slh.groupby("gender")["internet_usage"].mean()
```

```python
categories = avg_internet.index
values = avg_internet.values

plt.bar(categories, values, color="orange", width=0.5)

plt.xlabel("Gender")
plt.ylabel("Average Internet Usage")
plt.title("Average Internet Usage by Gender")

plt.show()
```
